```python
from collections import Counter
import csv

archivo = "autotagging_moodtheme.tsv"  # cambia por la ruta de tu archivo

tags_counter = Counter()

with open(archivo, newline='', encoding="utf-8") as f:
    reader = csv.reader(f, delimiter="\t")
    header = next(reader)  # saltar encabezado

    for fila in reader:
        # A partir de la columna TAGS (índice 5) hasta el final hay tags
        tags = fila[5:]  # pueden venir 1 o varios separados por TAB
        for tag in tags:
            tag = tag.strip()
            if tag:
                tags_counter[tag] += 1

# Mostrar resultados ordenados por ocurrencia (desc)
for tag, count in tags_counter.most_common():
    print(f"{tag}\t{count}")
```
